```python
import matplotlib.pyplot as plt
```

```python
x=[1,2,3,4]
y=[10,20,30,40]
plt.plot(x,y)
plt.show()
```

```python
x=[1,2,3,4]
y=[10,20,30,40]
plt.bar(x,y)
plt.show()
```

```python
y=[10,20,30,40]
plt.hist(y)
plt.show()
```

```python
x=[1,2,3,4]
y=[10,20,30,40]
plt.scatter(x,y)
plt.show()
```

```python
labels=["A","B","C"]
sizes =[30,40,30]
plt.pie(sizes,labels=labels,autopct="%1.1f%%")
plt.show()
```

```python
plt.plot(x,y,linestyle="--",marker="o")
plt.grid()
plt.show()
```
